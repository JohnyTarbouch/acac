```python
def merge_seeds(data_, num_bucket, bucket_size, name):
    def bucketing(data_, i):
        y = np.zeros(num_bucket)
        cur_bucket = 1
        count = 0
        for row in data_:
            ts, ret = row
            if ts >= total_step:
                break
            if ts < cur_bucket * bucket_size:
                y[cur_bucket-1] += ret
                count += 1
            else:
                y[cur_bucket-1] /= count
                count = 0
                cur_bucket += 1
        if cur_bucket < num_bucket:
            y[cur_bucket-1] /= count
            y[cur_bucket:] = y[cur_bucket-1]
            print(name, i,'Not enough timesteps, so copied the last')
        else:
            assert cur_bucket == num_bucket
            y[cur_bucket-1] /= count
        
        for _i in range(len(y)):
            if 0 < _i < len(y) - 1 and np.isnan(y[_i]):
                y[_i] = (y[_i-1] + y[_i+1]) / 2
        for _i in range(len(y)):
            if _i > 0 and np.isnan(y[_i]):
                y[_i] = y[_i-1]

        return y
    res = []
    for i,data__ in enumerate(data_):
        res.append(bucketing(data__,i))
    y = np.stack(res, axis=1)
    mean = y.mean(axis=1)
    stderr = y.std(axis=1)/np.sqrt(y.shape[1])
    return y, mean, stderr

def smooth(x,window_len=11,window='flat'):
    if window_len<3:
        return x
    s=np.r_[x[window_len-1:0:-1],x,x[-2:-window_len-1:-1]]
    if window == 'flat': #moving average
        w=np.ones(window_len,'d')
    else:
        w=eval('np.'+window+'(window_len)')

    y=np.convolve(w/w.sum(),s,mode='valid')
    return y

final_perf = {}
window_size = 2
for bench_i, benchmark in enumerate(benchmarks):
    print('##########', benchmark)
    final_perf[benchmark] = {}
    # LOAD DATA: assume   {model}_{benchmark}_{seed}.csv

    if 'bp' in benchmark:
        total_step = int(int(1e6) *0.5)
        num_bucket = 25
    elif 'ovc' in benchmark:
        if benchmark == "ovcABC7":
            total_step = 60 * int(1e6)  # 300k eps * 200 steps
        if '11' in benchmark:
            total_step = 100 * int(1e6)
        else:
            if 'R' in benchmark:
                total_step = 40 * int(1e6)
            else:
                total_step = 20 * int(1e6)
        num_bucket = 100
    x = np.arange(num_bucket)
    bucket_size = total_step // num_bucket  # 50000


    parent_dir = os.path.join(root_dir, benchmark)
    print(parent_dir)
    if not os.path.exists(parent_dir):
        print(parent_dir + ' not exist')
        continue
    data = {}
    available_names = []
    for name in names:
        method_dir = os.path.join(parent_dir, name)
        if not os.path.exists(method_dir):
            print(method_dir + ' not exist, skip')
            continue
        seed_files = sorted(
            [fname for fname in os.listdir(method_dir) if fname.endswith('.csv')],
            key=lambda fname: int(os.path.splitext(fname)[0])
        )
        if not seed_files:
            print(method_dir + ' has no seed csv, skip')
            continue
        data[name] = []
        for seed_file in seed_files:
            fname = os.path.join(method_dir, seed_file)
            with open(fname, 'r') as f:
                data[name].append(np.float32([i for i in csv.reader(f)][1:])[...,[0,-1]])
        available_names.append(name)
    if not available_names:
        print('No available methods for', benchmark)
        continue

    # PROCESS DATA
    final_data = {}
    for name in available_names:
        y, mean, stderr = merge_seeds(data[name], num_bucket, bucket_size, name)
        final_data[name+'_mean'] = smooth(mean, window_size)[:-window_size+1 if window_size > 3 else None]
        final_data[name+'_std'] = smooth(stderr, window_size)[:-window_size+1 if window_size > 3 else None]
    data = final_data

    # PLOT
    fig = plt.figure(figsize=(20,12))
    plt.rcParams.update({'font.size':56})
    plt.title(titles[benchmark])

    plt.axhline(y=0, color='black')
    for name in available_names:
        plt.plot(x, data[f'{name}_mean'],
                 label=plot_configs[name]["label"],
                 color=plot_configs[name]["color"],
                 linestyle=plot_configs[name]["linestyle"],
                 linewidth=4.0
                )
        plt.fill_between(x, data[f'{name}_mean']+data[f'{name}_std'], data[f'{name}_mean']-data[f'{name}_std'], color=plot_configs[name]["color"], alpha=0.1)
        final_perf[benchmark][name] = f"{np.round(np.mean(data[f'{name}_mean'][-5:]), 2)} \pm {np.round(np.mean(data[f'{name}_std'][-5:]), 2)}"
    plt.xlabel("Environment steps (1M)")
    plt.ylabel("Return")

    xtick_scale = total_step / 1e7
    if xtick_scale > 1:
        xtick_scale = int(xtick_scale)
    print(xtick_scale)
    plt.xticks([i for i in range(0, num_bucket, num_bucket//5)]+[num_bucket-1], [round(i,1) for i in np.arange(0, 10+1, step=2)*xtick_scale])
    
    if len(benchmarks) >= 2:
        if bench_i % 3 == 0:
            plt.ylabel("Return")
            
    plt.tight_layout()
    plt.savefig(f'../icml25_result/figure/main/{benchmark}_acac_user_only.pdf')




# print final performance
def print_final_perf(final_perf):
    df = pd.DataFrame.from_dict(final_perf, orient='index')

    print('=== Final Performance ===')
    print(df.round(2))

    latex_table = df.to_latex(float_format="%.2f", na_rep="--", index=True)
    print("\n=== LaTeX Table ===")
    print(latex_table)

print_final_perf(final_perf)




```

```python
import os
import csv
import numpy as np
import scipy
import matplotlib.pyplot as plt
import pandas as pd

names = ["acac_user", "acac_user_100M_g099_len200"]
benchmarks = ["ovcA7", "ovcC7", "ovcCR11_N6", "ovcABC7", "ovcABCR7"]

root_dir = "../icml25_result/log"

titles = {
    "bp6": "BoxPushing-6x6", "bp8": "BoxPushing-8x8", "bp10": "BoxPushing-10x10",
    "ovcA7": "Overcooked-A", "ovcB7": "Overcooked-B", "ovcC7": "Overcooked-C",
    "ovcAR7": "Overcooked-Rand-A", "ovcBR7": "Overcooked-Rand-B", "ovcCR7": "Overcooked-Rand-C",
    "ovcA11_N6": "Overcooked-Large-A", "ovcB11_N6": "Overcooked-Large-B", "ovcC11_N6": "Overcooked-Large-C",
    "ovcAR11_N6": "Overcooked-Large-Rand-A", "ovcBR11_N6": "Overcooked-Large-Rand-B", "ovcCR11_N6": "Overcooked-Large-Rand-C",
    "ovcABC7": "Overcooked-Random-ABC",
    "ovcABCR7": "Overcooked-Random-ABC-RandStart",

}

plot_configs = {
    "acac": {"label": "ACAC", "color": "tab:red", "linestyle": "-"},
    "acac_user": {"label": "ACAC User", "color": "tab:purple", "linestyle": "-."},
    "acac_user_100M_g099_len200": {"label": "ACAC User 100M", "color": "tab:orange", "linestyle": "-."},
    "acac_vanilla": {"label": "ACAC-Vanilla", "color": "tab:cyan", "linestyle": "-"},
    "mac_niacc": {"label": "Mac-NIACC", "color": "tab:brown", "linestyle": "-"},
    "mac_iaicc": {"label": "Mac-IAICC", "color": "tab:green", "linestyle": "-"},
    "acac_micro": {"label": r"$\mathrm{ACAC}_{\mathrm{micro}}$", "color": "tab:red", "linestyle": "--"},
    "acac_vanilla_micro": {"label": r"$\mathrm{ACAC}$-$\mathrm{Vanilla}_{\mathrm{micro}}$", "color": "tab:cyan", "linestyle": "--"},
    "niacc_micro": {"label": r"$\mathrm{NIACC}_{\mathrm{micro}}$", "color": "tab:green", "linestyle": "--"},
    "iaicc_micro": {"label": r"$\mathrm{IAICC}_{\mathrm{micro}}$", "color": "tab:brown", "linestyle": "--"},
    "mappo": {"label": r"$\mathrm{MAPPO}_{\mathrm{micro}}$", "color": "tab:olive", "linestyle": "--"},
    "acac_duplicate": {"label": "ACAC-Duplicate", "color": "tab:purple", "linestyle": "-"},
    "acac_micro_gae": {"label": "ACAC (micro-level GAE)", "color": "tab:green", "linestyle": "-"},
    "acac_eps0.005": {"label": r"ACAC ($\varepsilon$=0.005)", "color": "tab:green", "linestyle": "-"},
    "acac_eps0.015": {"label": r"ACAC ($\varepsilon$=0.015)", "color": "tab:blue", "linestyle": "-"},
    "acac_no_te": {"label": "ACAC (No TE)", "color": "tab:blue", "linestyle": "-"},
    "acac_no_sa": {"label": "ACAC (No SA)", "color": "tab:green", "linestyle": "-"},
}

def benchmark_steps(benchmark):
    if "bp" in benchmark:
        return int(0.5e6), 25
    if benchmark == "ovcABC7":
        return int(20e6), 100
    if benchmark == "ovcABCR7":
        return int(20e6), 100

    if "ovc" in benchmark:
        if "11" in benchmark:
            return int(100e6), 100
        if "R" in benchmark:
            return int(40e6), 100
        return int(20e6), 100
    raise ValueError(f"Unknown benchmark: {benchmark}")

def merge_seeds(data_, num_bucket, bucket_size, total_step, name):
    def bucketing(seed_data, seed_idx):
        y = np.full(num_bucket, np.nan, dtype=np.float32)

        for bucket_idx in range(num_bucket):
            lo = bucket_idx * bucket_size
            hi = (bucket_idx + 1) * bucket_size

            rows = seed_data[(seed_data[:, 0] >= lo) & (seed_data[:, 0] < hi)]
            if len(rows) > 0:
                y[bucket_idx] = np.mean(rows[:, 1])

        if np.isnan(y).any():
            print(name, seed_idx, "has unfinished/missing buckets; leaving them blank")

        return y

    per_seed = [bucketing(seed_data, i) for i, seed_data in enumerate(data_)]
    y = np.stack(per_seed, axis=1)

    mean = np.nanmean(y, axis=1)
    valid_count = np.sum(~np.isnan(y), axis=1)
    stderr = np.nanstd(y, axis=1) / np.sqrt(np.maximum(valid_count, 1))
    stderr[valid_count == 0] = np.nan

    return y, mean, stderr

def smooth_nan(x, window_size=2):
    if window_size < 3:
        return x

    s = pd.Series(x)
    return s.rolling(window_size, min_periods=1, center=True).mean().to_numpy()

final_perf = {}
window_size = 2

for bench_i, benchmark in enumerate(benchmarks):
    print("##########", benchmark)
    final_perf[benchmark] = {}

    total_step, num_bucket = benchmark_steps(benchmark)
    bucket_size = total_step // num_bucket
    x = ((np.arange(num_bucket) + 1) * bucket_size) / 1e6

    parent_dir = os.path.join(root_dir, benchmark)
    print(parent_dir)

    if not os.path.exists(parent_dir):
        print(parent_dir + " not exist")
        continue

    data = {}
    available_names = []

    for name in names:
        method_dir = os.path.join(parent_dir, name)
        if not os.path.exists(method_dir):
            print(method_dir + " not exist, skip")
            continue

        seed_files = sorted(
            [fname for fname in os.listdir(method_dir) if fname.endswith(".csv")],
            key=lambda fname: int(os.path.splitext(fname)[0])
        )

        if not seed_files:
            print(method_dir + " has no seed csv, skip")
            continue

        data[name] = []
        for seed_file in seed_files:
            fname = os.path.join(method_dir, seed_file)
            with open(fname, "r") as f:
                rows = np.float32([row for row in csv.reader(f)][1:])[:, [0, -1]]
                data[name].append(rows)

        available_names.append(name)

    if not available_names:
        print("No available methods for", benchmark)
        continue

    final_data = {}
    for name in available_names:
        y, mean, stderr = merge_seeds(data[name], num_bucket, bucket_size, total_step, name)
        final_data[name + "_mean"] = smooth_nan(mean, window_size)
        final_data[name + "_std"] = smooth_nan(stderr, window_size)

    fig = plt.figure(figsize=(20, 12))
    plt.rcParams.update({"font.size": 56})
    plt.title(titles.get(benchmark, benchmark))
    plt.axhline(y=0, color="black")

    for name in available_names:
        mean = final_data[name + "_mean"]
        std = final_data[name + "_std"]
        mask = ~np.isnan(mean)

        plt.plot(
            x[mask],
            mean[mask],
            label=plot_configs[name]["label"],
            color=plot_configs[name]["color"],
            linestyle=plot_configs[name]["linestyle"],
            linewidth=4.0,
        )

        plt.fill_between(
            x[mask],
            mean[mask] + std[mask],
            mean[mask] - std[mask],
            color=plot_configs[name]["color"],
            alpha=0.1,
        )

        last_values = mean[mask][-5:]
        last_stds = std[mask][-5:]
        final_perf[benchmark][name] = (
            f"{np.round(np.nanmean(last_values), 2)} "
            f"\\pm {np.round(np.nanmean(last_stds), 2)}"
        )

    plt.xlabel("Environment steps (1M)")
    plt.ylabel("Return")
    plt.xlim(0, total_step / 1e6)
    plt.xticks(np.linspace(0, total_step / 1e6, 6))
    plt.legend()
    plt.tight_layout()

    os.makedirs("../icml25_result/figure/main", exist_ok=True)
    plt.savefig(f"../icml25_result/figure/main/{benchmark}_acac_user_only.pdf")
    plt.show()

def print_final_perf(final_perf):
    df = pd.DataFrame.from_dict(final_perf, orient="index")
    print("=== Final Performance ===")
    print(df)

    latex_table = df.to_latex(na_rep="--", index=True)
    print("\n=== LaTeX Table ===")
    print(latex_table)

print_final_perf(final_perf)
```
